# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

```python
np.random.seed(0)
ts1 = 100 * (np.random.normal(size=numdates) / 5 + np.arange(numdates) / 16)
np.random.seed(1)
ts2 = 100 * (1 - (np.random.normal(size=numdates) / 8 + np.arange(numdates) / 16))
dates = pd.date_range(start='2023/01/01', end='2024/01/01', freq='MS')[:numdates]

fig, ax = plt.subplots(figsize=(10, 5))

sns.lineplot(x=dates[:-3], y=ts1[:-3], color='lightgrey', linewidth=2, ax=ax)
sns.lineplot(x=dates[:-3], y=ts2[:-3], color='lightgrey', linewidth=2, linestyle='--', ax=ax)

sns.lineplot(x=dates[-4:], y=ts1[-4:], color='steelblue', linewidth=3, label='Nitrate', ax=ax)
sns.lineplot(x=dates[-4:], y=ts2[-4:], color='coral', linewidth=3, linestyle='--', label='Phosphate', ax=ax)

ax.annotate('Focus: Last 3 months', xy=(dates[-3], ts1[-3]),
            xytext=(dates[-6], ts1[-3] + 15),
            arrowprops=dict(arrowstyle='->', color='steelblue'),
            fontsize=10, color='steelblue')

ax.set_title('Nitrate and Phosphate Levels (Last 3 Months Highlighted)')
ax.set_xlabel('Year and month')
ax.legend()
plt.tight_layout()
plt.show()
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
data = {
    'Communication': [4, 3, 2, 5, 3],
    'Work-Life Balance': [3, 2, 4, 4, 2],
    'Career Growth': [2, 3, 3, 4, 1],
    'Management': [4, 4, 2, 5, 3],
    'Compensation': [3, 2, 2, 4, 2]
}

departments = ['Engineering', 'Marketing', 'Sales', 'Finance', 'Operations']
df = pd.DataFrame(data, index=departments)

fig, ax = plt.subplots(figsize=(8, 5))

sns.heatmap(df, annot=True, fmt='d', cmap='RdYlGn', vmin=1, vmax=5,
            linewidths=0.5, ax=ax, cbar_kws={'label': 'Satisfaction (1-5)'})

ax.set_title('Employee Satisfaction by Department and Category', fontsize=13, fontweight='bold')
ax.set_xlabel('')
ax.set_ylabel('')
plt.tight_layout()
plt.show()
```

I chose a heatmap because it allows the viewer to instantly identify patterns across two dimensions — department and satisfaction category without needing to read individual numbers. The green/red color scale uses pre-attentive attributes to draw attention to problem areas (red = low satisfaction) immediately, which aligns with the SWD principle of using color intentionally to highlight what matters.
